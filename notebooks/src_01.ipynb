# SÉRIES TEMPORAIS

Início das aquisições: 01/01/2022 11:00:00<br>
Fim das aquisições: 30/12/2022 11:00:00<br>
<br>
Cada aquisição é feita a cada 10 minutos.


```python
# Converte a primeira coluna para datetime
data.iloc[:, 0] = pd.to_datetime(data.iloc[:, 0])

# Eixo X = data/hora real das medições
tempo = data.iloc[:, 0]
```

```python
# Todas as colunas menos a primeira
cols = data.columns[1:]

# Grupos de 8 variáveis
cols_groups = [cols[i:i+8] for i in range(0, len(cols), 8)]
```

```python
for group in cols_groups:

    fig, axs = plt.subplots(
        len(group),
        1,
        figsize=(14, 2 * len(group)),
        sharex=True
    )

    if len(group) == 1:
        axs = [axs]

    for i, col in enumerate(group):

        ax = axs[i]

        # Todos os pontos da série temporal
        ax.plot(tempo, data[col])

        ax.set_title(col)
        ax.set_ylabel("Valor")
        ax.grid(True)

    axs[-1].set_xlabel("Data / Hora")

    
    plt.tight_layout()
    plt.show()
```

**Matriz de Correlação MENSAL**


```python
data['date-time'] = pd.to_datetime(data['date-time'])

#data inicial
t0 = pd.Timestamp("2022-01-01 11:00:00") #inicio do mes 1

data['mes_30d'] = ((data['date-time'] - t0).dt.days // 30) + 1    #df["date-time"] - t0 --> calcula quantos dias se passaram desde 01/01/2022
                                                                  # .dt.days --> só pega números inteiros

#verificar se cada mes tem 30 dias mesmo
data.groupby("mes_30d")["date-time"].agg(["min", "max"])
```

```python
#verificando quantos dados tem em cada mês
data.groupby("mes_30d")["date-time"].count()
```

```python
#criando os datasets mensais
mensal = {}

for i in range(1, 14):
   mensal[f'mes_{i}'] = data[data['mes_30d'] == i ].copy()
```

```python
correlacoes = {}        #dicionario das correlações
```

```python
for i in range(1,14):
    df_mes = mensal[f'mes_{i}'].drop(columns=['date-time', 'mes_30d'])
    df_mes = df_mes.loc[:, df_mes.nunique() > 1]

    corr = df_mes.corr()
    
    correlacoes[f'corr_mes_{i}'] = corr
```

```python
#plot tirando as variaveis que estão sempre constantes (sempre zero)

for i in range(1, 14):
    corr = correlacoes[f"corr_mes_{i}"]
    
    plt.figure(figsize=(22,18))
    sns.heatmap(corr, cmap="coolwarm", center=0, annot=True, fmt=".2f")
    plt.title(f"Matriz de Correlação - Mês {i}")
    plt.show()
```

Matriz de correlação da variável alvo com as variáveis preditoras

```python
# matriz de correlação da variável alvo com as variáevis preditoras

target = "364602_31231_FI_305"         # variável alvo

corr_target = data.select_dtypes(include="number").corr()[[target]]

plt.figure(figsize=(20, 5))
plt.title(f"Matriz de correlação em relação à variável '{target}'")
sns.heatmap(corr_target.T, annot=True, cmap='coolwarm')
plt.show()
```

Matriz de correlação **mensal** da **variável alvo**.

```python
correlacoes_target = {} 

for i in range(1,14):
    df_mes_relev = mensal[f'mes_{i}'].drop(columns=['date-time', 'mes_30d'])
    df_mes_relev = df_mes_relev.loc[:, df_mes_relev.nunique() > 1] #tira os espaços em branco das variáveis com valores cte (zero)

    if target not in df_mes_relev.columns:
        print(f"Mês {i}: variável alvo não existe")
        continue

    #correlação com a variável alvo
    corr_target = df_mes_relev.corr()[[target]]   #dupla chave = DataFrame
    
    correlacoes_target[f'corr_mes_{i}'] = corr_target
```

```python
for i in range(1,14):
    if f"corr_mes_{i}" not in correlacoes_target:
        continue
    
    corr = correlacoes_target[f"corr_mes_{i}"]

    plt.figure(figsize=(20,4))
    sns.heatmap(corr.T, cmap="coolwarm", center=0, annot=True, fmt=".2f")
    plt.title(f"Correlação com a variável: '{target}' - Mês {i}")
    plt.show()
```

```python
top_corr_target_mes = {} #dicionário com as correlações

for i in range(1, 14):
    df_mes_relev = mensal[f'mes_{i}'].drop(columns=['date-time', 'mes_30d'])
    
    #remove variáveis constantes
    #df_mes_relev = df_mes_relev.loc[:, df_mes_relev.nunique() > 1]
    
    #garante que a variável existe
    if target not in df_mes_relev.columns:
        print(f"Mês {i}: {target} não existe")
        continue
    
    #correlação com a variável alvo
    corr_target = df_mes_relev.corr()[target]
    
    #remove ela mesma (variável target)
    corr_target = corr_target.drop(target)
    
    #módulo da correlação
    corr_target_abs = corr_target.abs().sort_values(ascending=False)
    
    #guarda top 25
    top_corr_target_mes[f"mes_{i}"] = corr_target_abs.head(25)
```

```python
for i in range(1,14):
    print(f"\n===== MÊS {i} =====")
    print(top_corr_target_mes[f"mes_{i}"])
```

**MÊS X CORRELAÇÃO**

Cada gráfico representa um mês e seus valores com as suas respectivas variáveis.

```python
df_corr_mes = pd.DataFrame()              #dataframe vazio --> vira tabela onde cada coluna é um mês e cada linha uma variavel

for i in range(1,14):
    variavel = top_corr_target_mes[f"mes_{i}"]
    df_corr_mes[f"mes_{i}"] = variavel
```

```python
#gráfico da correlação de cada variavel em cada mês
for i in range(1, 14):
    
    if f"mes_{i}" not in top_corr_target_mes: #verifica se o mês existe
        continue
    
    variavel = top_corr_target_mes[f"mes_{i}"]
    
    plt.figure(figsize=(16,2))
    # gráfico de pontos
    plt.plot(range(len(variavel)), variavel.values, linestyle='-', marker='o', markersize=4, linewidth=1, label='corr')
    
    # nomes dos sensores no eixo x
    plt.xticks(range(len(variavel)), variavel.index, rotation=90)
    plt.ylabel("módulo correlação")
    plt.xlabel("")
    plt.title(f"Correlações com '{target}' - Mês {i}")
    plt.ylim(0,1)
    plt.legend()
    plt.show()

plt.show()
```

Gráfico do módulo da maior correlação a cada mês

```python
meses = []           #guarda os meses
max_corr = []        #guarda os maiores valores da correlação

for i in range(1, 13):
    if f"mes_{i}" not in top_corr_target_mes:                  #verifica se o mês existe
        continue
    
    meses.append(i) #guarda o mês
    max_corr.append(top_corr_target_mes[f"mes_{i}"].iloc[0])   #pega a maior correlaçao do mês e guarda
```

```python
plt.figure(figsize=(20,3))
plt.plot(meses, max_corr, marker="o", linestyle='-', markersize=4, linewidth=1, label='corr') 
plt.xlabel("Mês (janela de 30 dias)")
plt.ylabel("Valor da correlção")
plt.title("MÊS X CORR")
#plt.ylim(0,1)   #correlação vai de 0 a 1
plt.legend()
plt.show()
```
